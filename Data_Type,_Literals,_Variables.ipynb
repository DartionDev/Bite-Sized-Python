## Part A — Built-in types and type()

Five simple types: `int`, `float`, `bool`, `str`, `complex`. Four collections: `list`, `tuple`, `set`, `dict`.
`bool` is a subclass of `int`. Floats are binary approximations, never compare them with `==`.

```python

age = 25; price = 19.99; is_active = True; name = "Darshan"; z = 3 + 4j
nums = [1, 2, 3]; point = (10, 20); tags = {"py", "ai"}; user = {"name": "Darshan", "age": 25}

for v in (age, price, is_active, name, z, nums, point, tags, user):
    print(type(v).__name__, "->", v)
```

```python

import math
print(0.1 + 0.2 == 0.3)              # False, classic float trap
print(math.isclose(0.1 + 0.2, 0.3))  # True, the correct way to compare
print(isinstance(True, int))          # True, bool IS an int
print(True + True)                    # 2
```

## Part B — Literals: many notations, one value

`0b1010`, `0o310`, `0x12C`, and `100`-style decimal all compile to plain `int` objects, notation is not remembered at runtime. Floats support scientific notation (`1.5e2`). Raw strings (`r"..."`) turn off backslash escaping.

```python

a, b, c, d = 0b1010, 100, 0o310, 0x12C
print(a, b, c, d)                 # 10 100 200 300, four notations, real ints

f1, f2, f3 = 10.5, 1.5e2, 1.5e-3
print(f1, f2, f3)                 # 10.5 150.0 0.0015

raw = r"this is a new line: \n, not an actual line break"
print(raw)

print(True + 4, False + 10)       # 5 10
```

## Part C — Variables are names bound to objects

`b = a` copies a *reference*, not the value. Mutating through one name shows through every name pointing at the same object. Rebinding a name never affects other names pointing at the old object.

```python

a = [1, 2, 3]
b = a
b.append(4)
print(a)              # [1, 2, 3, 4]  <- a changed too, same object

n = 10
m = n
m = 20                 # rebinding, not mutation
print(n, m)             # 10 20, n is untouched

first, *rest = [10, 20, 30, 40]
print(first, rest)      # 10 [20, 30, 40]
```

## Part D — Keywords and identifier rules

`True`, `False`, `None` are capitalized keywords. Identifiers can't start with a digit, can't be a keyword, and are case-sensitive. `print`, `list`, `str`, `type` are built-ins, not keywords, they *can* be shadowed, which is a common source of confusing bugs.

```python

import keyword
print(keyword.iskeyword("class"), keyword.iskeyword("print"))   # True False
print(len(keyword.kwlist))
```

```python
original = [1, 2, 3]
alias = original
copy_of = original.copy()
alias.append(99)
print(original)   # [1, 2, 3, 99]
print(alias)       # [1, 2, 3, 99]
print(copy_of)      # [1, 2, 3], unaffected
# alias and original are two names for the SAME list object, so mutating through
# alias is visible through original too. copy_of is a separate object.
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def describe(x):
    return f"{type(x).__name__}: {x}"

print(describe(0b1111))          # int: 15
print(describe(6.022e23))         # float: 6.022e+23
print(describe(2 + 3j))            # complex: (2+3j)
print(describe(r"C:\new\path"))     # str: C:\new\path
```
