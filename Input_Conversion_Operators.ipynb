## Part A — input() always returns a string

`input()` reads one line from stdin and always hands back a `str`. Convert explicitly with `int()`/`float()` before doing math. Implicit conversion only happens between compatible numeric types.

```python

n1 = "5"    # pretend this came from input("Enter a number: ")
n2 = "7"
print(n1 + n2)          # '57'  string concatenation, not addition!
print(float(n1) + float(n2))   # 12.0, correct after explicit conversion
```

```python

print(5 + 5.5)         # 10.5  implicit int -> float promotion
print(int(4.9), int(-4.9))     # 4 -4, truncates toward zero
print(round(4.5), round(5.5))  # 4 6, banker's rounding
print(bool("False"))    # True!! any non-empty string is truthy
```

## Part B — Six operator categories

Arithmetic (`+ - * / // % **`), relational (`== != < <= > >=`), logical (`and or not`), bitwise (`& | ^ ~ << >>`), assignment (`= += -= ...`), membership (`in`, `not in`).

`//` **floors** toward negative infinity, it does not just truncate. `and`/`or` return an operand, not necessarily a bool.

```python

print(4 / 2, 4 // 2, 4 % 3, 5 ** 2)     # 2.0 2 1 25
print(-7 // 2, -7 % 3)                    # -4 2, floors toward -infinity
print(1 and 0, 1 or 0, not 1)              # 0 1 False
print(2 & 3, 2 | 3, 2 ^ 3)                  # 2 3 1
```

```python

# Digit-splitting exercise from the post
number = 345
c = number % 10; number //= 10
b = number % 10; number //= 10
a = number % 10
print(a + b + c)   # 12
```

### Exercise 6: Digit sum for any length number  
*Difficulty: Intermediate*

Your post's digit-splitting trick assumes exactly 3 digits. Write a `digit_sum(n)` function using `divmod` in a loop that works for any non-negative integer, and test it on `345`, `7`, and `98765`.

```python
def digit_sum(n):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def digit_sum(n):
    total = 0
    while n:
        n, digit = divmod(n, 10)
        total += digit
    return total

print(digit_sum(345), digit_sum(7), digit_sum(98765))
```

### Exercise 10: Mini project: simple 4-function calculator  
*Difficulty: Challenge*

Write `calculate(a, b, op)` where `op` is one of `'+' '-' '*' '/'` as a string, returning the result. Divide by zero should return the string `"Error: division by zero"` instead of crashing. Test all four operators plus a divide-by-zero case.

```python
def calculate(a, b, op):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def calculate(a, b, op):
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        if b == 0:
            return "Error: division by zero"
        return a / b
    else:
        return f"Unknown operator: {op}"

print(calculate(6, 3, "+"))
print(calculate(6, 3, "-"))
print(calculate(6, 3, "*"))
print(calculate(6, 3, "/"))
print(calculate(6, 0, "/"))
```
