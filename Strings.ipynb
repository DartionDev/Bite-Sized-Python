## Recap

A `str` is an **immutable** sequence of Unicode characters. Indexing (`s[0]`) reads one character, slicing (`s[start:stop:step]`) reads a range and always returns a *new* string. Every string method returns a new string too, `s.upper()` never changes `s` itself. `+` concatenates, `*` repeats, comparisons are lexicographic by code point.

```python

s = "welcome to python"
print(s[0], s[-1])          # w n
print(s[:7], s[7:], s[::-1])  # welcome ' to python' nohtyp ot emoclew
print(s.upper(), "|", s)      # UPPER version printed, s itself unchanged
```

```python

s = "my name is darshan"
print(s.count("a"), s.find("z"), s.startswith("my"))   # 3 -1 True
words = s.split()
print(words, "-".join(words))
```

```python

try:
    s = "immutable"
    s[0] = "I"
except TypeError as e:
    print("TypeError:", e)
```

### Exercise 7: Palindrome checker, properly normalized  
*Difficulty: Intermediate*

Write `is_palindrome(text)` that ignores case, spaces, and punctuation (keep only alphanumeric characters, casefolded) before comparing. Test on `"A man, a plan, a canal: Panama"` (True) and `"hello"` (False).

```python
def is_palindrome(text):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def is_palindrome(text):
    cleaned = "".join(ch.casefold() for ch in text if ch.isalnum())
    return cleaned == cleaned[::-1]

print(is_palindrome("A man, a plan, a canal: Panama"))
print(is_palindrome("hello"))
```

### Exercise 10: Title-case a sentence, the correct way  
*Difficulty: Challenge*

`.title()` mishandles apostrophes (`"they're".title()` gives `"They'Re"`). Write `smart_title(sentence)` that capitalizes only the first letter of each whitespace-separated word, leaving the rest of each word untouched. Test on `"they're going to python's world"`.

```python
def smart_title(sentence):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def smart_title(sentence):
    words = sentence.split(" ")
    capitalized = [w[:1].upper() + w[1:] for w in words if w]
    return " ".join(capitalized)

print(smart_title("they're going to python's world"))
```

*Difficulty: Challenge*

Write `caesar_shift(text, shift)` that shifts each lowercase letter forward by `shift` positions in the alphabet (wrapping z back to a), leaving non-letters unchanged. Use `ord`/`chr`. Test with `caesar_shift("hello world", 3)` (should give `"khoor zruog"`).

```python
def caesar_shift(text, shift):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def caesar_shift(text, shift):
    result = []
    for ch in text:
        if ch.isalpha():
            base = ord("a")
            result.append(chr((ord(ch) - base + shift) % 26 + base))
        else:
            result.append(ch)
    return "".join(result)

print(caesar_shift("hello world", 3))
```
