## Recap

`import x` checks `sys.modules` (cache) first, then finds and executes the module file once, caching it. Module code only runs once per process no matter how many times you import it. The standard library ships with Python (`math`, `random`, `datetime`, ...). Third-party packages come from PyPI via `pip`. In Colab, `!pip install` or the `%pip install` magic both work, since cells run in a real shell attached to the notebook's Python.

```python

import math, random, datetime
print(math.sqrt(196), math.factorial(4))
print(random.randint(1, 100))
print(datetime.datetime.now())
```

```python

import sys
print("math" in sys.modules)                 # True once imported
print(sys.modules["math"] is math)             # True, same cached object
print(sys.path[:2])                            # where Python looks, in order
```

```python

# Installing a third-party package in Colab
%pip install -q requests
import requests
print(requests.__version__)
```

```python
def inspect_module(mod):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def inspect_module(mod):
    file_ = getattr(mod, "__file__", "(built-in, no file)")
    public = [n for n in dir(mod) if not n.startswith("_")]
    print(f"{mod.__name__}: file={file_}, public_names={len(public)}")

import math, random
inspect_module(math)
inspect_module(random)
```
