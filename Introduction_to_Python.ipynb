## Part A — What actually runs your code

- Python is the *language spec*. **CPython** is the C program most people mean when they say "Python".
- CPython compiles your source to **bytecode**, then an evaluation loop executes that bytecode. It's not pure interpretation and not full compilation to machine code.
- `sys.executable` tells you exactly which interpreter is running right now, the fix for 90% of "pip installed it but import fails" bugs is confirming this matches the `pip` you used.
- Colab gives you a fresh CPython + a shell you can reach with `!` (or the `%pip`/`%%bash` magics).

```python

import sys
print("Python version:", sys.version)
print("Interpreter path:", sys.executable)
print("Platform:", sys.platform)
```

```python

# Peek at the bytecode CPython generates for a tiny expression
import dis
dis.dis("x = 1 + 2")
```

## Part B — print(): sep, end, file, flush

`print(*objects, sep=' ', end='\n', file=None, flush=False)`

- Converts every object with `str()`, joins them with `sep`, and finishes with `end`.
- `file` defaults to `sys.stdout`, looked up *at call time*, which is why redirecting `sys.stdout` works and why `file=sys.stderr` sends a message to the error stream instead.
- Returns `None`. Printing is not returning.

```python

name, age = "Darshan", 25
print("Name:", name, "Age:", age)          # default sep=' '
print("2026", "09", "24", sep="-")          # custom sep
print("Loading", end="...")
print("done")                                # end='' kept us on the same line
print(*["a", "b", "c"], sep=", ")            # unpack a list into print's arguments
```

```python

import sys
print("this goes to the error stream", file=sys.stderr)

import io
buf = io.StringIO()
print("captured!", file=buf)
print(repr(buf.getvalue()))
```
