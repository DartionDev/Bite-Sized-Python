## Recap

`while` repeats while a condition holds, use it when you don't know the number of rounds up front. `for x in iterable` walks any iterable via the iterator protocol, `range(start, stop, step)` is a lazy sequence with `stop` excluded. Nested loops run the inner loop to completion for every outer pass. `break` exits the innermost loop, `continue` skips to the next iteration, `pass` is a no-op placeholder. A loop's `else` runs only if the loop finished *without* hitting `break`.

```python

i = 0
while i < 5:
    print(i)
    i += 1
```

```python

for i in range(1, 10, 2):
    print(i)          # 1 3 5 7 9

for i in range(1, 5):
    for j in range(1, 5):
        pass           # inner loop runs fully for every outer i
print("16 total inner passes happened above")
```

```python

# Prime check using for/else -- else runs only if break never fired
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False
    return True

print([n for n in range(2, 20) if is_prime(n)])
```

### Exercise 8: Number-guessing game (deterministic test version)  
*Difficulty: Intermediate*

Write a function `guess_feedback(guesses, target)` that takes a list of guesses and a target int, returns the 1-based attempt number of the first correct guess, or `-1` if none matched, printing a hint (`"higher"`/`"lower"`) for every wrong guess along the way. Test with `guesses=[50, 75, 62, 68, 65]`, `target=65`.

```python
def guess_feedback(guesses, target):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def guess_feedback(guesses, target):
    for attempt, g in enumerate(guesses, start=1):
        if g == target:
            return attempt
        print("higher" if g < target else "lower")
    return -1

print(guess_feedback([50, 75, 62, 68, 65], 65))
```

### Exercise 9: Break out of nested loops with a function  
*Difficulty: Challenge*

`break` only exits the innermost loop. Write `find_pair(target)` that searches `i, j` both in `range(10)` for a pair where `i * j == target`, returning the first `(i, j)` tuple found (or `None`), using a `return` to escape both loops cleanly. Test with `target=42`.

```python
def find_pair(target):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def find_pair(target):
    for i in range(10):
        for j in range(10):
            if i * j == target:
                return (i, j)
    return None

print(find_pair(42))
```

### Exercise 11: Mini project: prime sieve up to N  
*Difficulty: Challenge*

Write `primes_up_to(n)` that returns a list of all primes from 2 up to `n` (inclusive), using nested loops with `break` and `for/else` (no importing a primality library). Test with `n=30`.

```python
def primes_up_to(n):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def primes_up_to(n):
    result = []
    for num in range(2, n + 1):
        for divisor in range(2, int(num ** 0.5) + 1):
            if num % divisor == 0:
                break
        else:
            result.append(num)
    return result

print(primes_up_to(30))
```

### Exercise 12: Population growth simulator  
*Difficulty: Challenge*

Starting at `population = 1000` growing 5% a year, use a `while` loop to find how many whole years it takes to exceed `2000`, printing the year and population each iteration, then the final answer.

```python
population = 1000
year = 0
# Your code here
```

**Solution** — try it yourself first, then run the cell below to check.

```python
population = 1000
year = 0
while population <= 2000:
    year += 1
    population *= 1.05
    print(year, round(population, 1))
print(f"Population exceeds 2000 after {year} years")
```
