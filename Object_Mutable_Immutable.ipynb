## Recap

Every object has an **identity** (`id(obj)`, unique and stable for its lifetime), a **type** (never changes), and a **value** (only mutable types can change this in place). The correct test for mutability is: does `id()` stay the same after the "change"? If yes, it mutated. If no, Python quietly created a new object and rebound your variable.

```python

sugar_amount = 2
print(id(sugar_amount))
sugar_amount = 12
print(id(sugar_amount))   # different id -- rebound to a NEW int object
```

```python

spice_mix = set()
print(id(spice_mix))
spice_mix.add("clove")
spice_mix.add("ginger")
print(id(spice_mix))       # SAME id -- mutated the existing set object
```

```python

def add_spice(spices):
    spices.add("clove")   # mutates the caller's object, they share one id

mix = set()
add_spice(mix)
print(mix)                 # {'clove'} -- caller sees the change

def bump(n):
    n = n + 1               # rebinds the LOCAL name, new int object
    return n

sugar = 2
bump(sugar)
print(sugar)                # 2, unaffected
```

### Exercise 8: Fix the mutable default argument bug  
*Difficulty: Intermediate*

Fix `add_item` from the previous exercise using the `None`-default pattern, so each call without an explicit bucket starts fresh.

```python
def add_item(item, bucket=None):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def add_item(item, bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket

print(add_item("a"))
print(add_item("b"))   # ["b"], not ["a", "b"]
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def track(obj):
    original_id = id(obj)
    def check():
        print(id(obj) == original_id)
    return check

data = [1, 2]
watch = track(data)
watch()               # True
data.append(3)
watch()               # True, still the same object
data = data + [4]     # rebinds the outer 'data' name, but track() closed over the OLD object
watch()               # True -- the closure still refers to the original list, id unchanged
print(data)            # [1, 2, 3, 4], a NEW list now
```
