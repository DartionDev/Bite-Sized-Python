## Recap

`if` runs a block when a condition is truthy. `elif` chains extra checks, only the *first* match runs and later conditions are never evaluated. `else` catches everything else. Indentation (not braces) marks the block. Separate `if`s are independent and can all fire; an `elif` chain guarantees at most one branch runs.

```python

ACTUAL_EMAIL, ACTUAL_PASSWORD = "darshan@buyweb.com", "1234"
email, password = "darshan@buyweb.com", "wrong"

if email == ACTUAL_EMAIL and password == ACTUAL_PASSWORD:
    print("Login successful")
elif email == ACTUAL_EMAIL:
    print("Please try again")
else:
    print("Invalid credentials")
```

```python

# Minimum of three -- watch the tie bug with strict <
def minimum_buggy(a, b, c):
    if a < b and a < c:
        return a
    elif b < a and b < c:
        return b
    else:
        return c

print(minimum_buggy(1, 1, 5))   # bug: prints 5 instead of 1 on a tie
```

```python

menu_choice = "3"
if menu_choice == "1":
    print("Pin change")
elif menu_choice == "2":
    print("Balance check")
elif menu_choice == "3":
    print("Withdraw")
elif menu_choice == "4":
    print("Deposit")
elif menu_choice == "5":
    print("Exit")
else:
    print("Invalid option")
```

### Exercise 9: Password strength checker  
*Difficulty: Challenge*

Given `pwd = "Abc123!!"`, print `"strong"` if it's at least 8 characters AND contains at least one digit AND at least one of `!@#$%^&*`, otherwise `"weak"`. Use `any()` with generator expressions for the digit/symbol checks.

```python
pwd = "Abc123!!"
# Your code here
```

**Solution** — try it yourself first, then run the cell below to check.

```python
pwd = "Abc123!!"
symbols = set("!@#$%^&*")
has_digit = any(ch.isdigit() for ch in pwd)
has_symbol = any(ch in symbols for ch in pwd)
if len(pwd) >= 8 and has_digit and has_symbol:
    print("strong")
else:
    print("weak")
```

```python
def judge(p1, p2):
    # Your code here
    pass
```

**Solution** — try it yourself first, then run the cell below to check.

```python
def judge(p1, p2):
    if p1 == p2:
        return "tie"
    beats = {"rock": "scissors", "scissors": "paper", "paper": "rock"}
    if beats[p1] == p2:
        return "p1"
    return "p2"

for opponent in ["rock", "paper", "scissors"]:
    print("rock vs", opponent, "->", judge("rock", opponent))
```
